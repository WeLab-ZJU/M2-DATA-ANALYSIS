```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import re 
from scipy.cluster import hierarchy
from scipy import cluster   
from scipy.stats import zscore
from scipy.signal import find_peaks
from sklearn.preprocessing import MinMaxScaler,StandardScaler
```

```python
path = r"${PROJECT_ROOT}\Figure 1\cFOS.csv"
data = pd.read_csv(path, index_col=None)
cfos_data = data.iloc[:,:-2]
IPS_data = data.iloc[:,-2:]
cfos_flat = cfos_data.values.flatten()
mean_cfos = np.mean(cfos_flat)
std_cfos = np.std(cfos_flat)
cfos_scaled = (cfos_data - mean_cfos) / std_cfos
```

```python
cfos_scaled.to_csv(r"${PROJECT_ROOT}\Figure 1_v2\cfos_scaled.csv", index=False)
```

```python
sns.heatmap(cfos_scaled, cmap='coolwarm')
plt.savefig(r"${PROJECT_ROOT}\Figure 1_v2\cFOS_heatmap.pdf", dpi=300)
```

```python
IPS_flat = IPS_data.values.flatten()
mean_IPS = np.mean(IPS_flat)
std_IPS = np.std(IPS_flat)
IPS_scaled = (IPS_data - mean_IPS) / std_IPS
sns.heatmap(IPS_scaled, cmap='coolwarm')
plt.savefig(r"${PROJECT_ROOT}\Figure 1_v2\IPS_heatmap.pdf", dpi=300)
```

```python
combined_scaled = pd.concat([cfos_scaled, IPS_scaled], axis=1)
combined_scaled.to_csv(r"${PROJECT_ROOT}\Figure 1_v2\Figure1_heatmap.csv", index=False)
```
